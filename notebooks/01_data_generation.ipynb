# Quantum Channel Noise Simulation
This notebook simulates dynamic non-linear noise (Phase & Amplitude Damping) in a quantum network channel and exports the dataset.

```python
import numpy as np
import pandas as pd
import os

def generate_quantum_channel_noise(n_steps=2000, dt=0.01, seed=42):
    """
    Simulates dynamic noise and decoherence rates (gamma_phase, gamma_amp)
    for a quantum channel over time using coupled non-linear stochastic processes.
    """
    np.random.seed(seed)
    t = np.linspace(0, n_steps * dt, n_steps)
    
    # Non-linear dynamic channel fluctuations (Phase Damping & Amplitude Damping)
    gamma_phase = 0.5 * np.sin(0.1 * t) + 0.3 * np.cos(0.05 * t) + 0.1 * np.random.normal(size=n_steps)
    gamma_amp = 0.4 * np.cos(0.08 * t) + 0.2 * np.sin(0.15 * t) + 0.08 * np.random.normal(size=n_steps)
    
    # Entanglement Fidelity metric estimation
    fidelity = 0.9 * np.exp(-0.2 * (np.abs(gamma_phase) + np.abs(gamma_amp))) + 0.02 * np.random.normal(size=n_steps)
    fidelity = np.clip(fidelity, 0.0, 1.0)
    
    df = pd.DataFrame({
        'time': t,
        'gamma_phase': gamma_phase,
        'gamma_amp': gamma_amp,
        'fidelity': fidelity
    })
    
    return df

# Create data directory and generate dataset
os.makedirs("../data", exist_ok=True)
df_noise = generate_quantum_channel_noise()
output_path = "../data/quantum_channel_noise.csv"
df_noise.to_csv(output_path, index=False)

print(f"[+] Quantum channel noise dataset generated successfully at: {output_path}")
df_noise.head()
```
